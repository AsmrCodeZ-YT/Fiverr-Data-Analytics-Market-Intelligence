```python
import json
from collections import Counter

# Read the Fiverr gigs JSON file from the UC volume
path = "/Volumes/demo/bronze/fiverr/all_gigs.json"
with open(path, "r", encoding="utf-8") as f:
    gigs = json.load(f)

print(f"Total gigs in file: {len(gigs)}")

# Fiverr "Top Rated Seller" indicators
def is_top_seller(gig):
    seller = gig.get("seller") or {}
    listing = gig.get("listing") or {}
    level = str(seller.get("level") or "").upper()
    level_raw = str(seller.get("level_raw") or "").upper()
    listing_level = str(listing.get("seller_level") or "").upper()
    return (
        "TOP_RATED" in level
        or "TOP_RATED" in level_raw
        or "TOP_RATED" in listing_level
    )

# Count gigs whose seller is a Top Rated Seller
top_gigs = [g for g in gigs if is_top_seller(g)]
print(f"Gigs offered by Top Rated Sellers: {len(top_gigs)}")

# Unique sellers (by username) and their levels
sellers = {}
for g in gigs:
    seller = g.get("seller") or {}
    username = seller.get("username")
    if username and username not in sellers:
        sellers[username] = str(seller.get("level") or "unknown")

print(f"Total unique sellers: {len(sellers)}")

# Unique Top Rated Sellers
top_sellers = [u for u, lvl in sellers.items() if "TOP_RATED" in lvl.upper()]
print(f"Unique Top Rated Sellers: {len(top_sellers)}")
print(f"Share of sellers who are Top Rated: {len(top_sellers) / len(sellers) * 100:.1f}%")

# Distribution of all seller levels for context
print("\nSeller level distribution (unique sellers):")
for lvl, count in Counter(sellers.values()).most_common():
    print(f"  {lvl}: {count}")

# Show the Top Rated Seller usernames
if top_sellers:
    display(spark.createDataFrame([(u,) for u in sorted(top_sellers)], ["top_rated_seller"]))
```
